```python
import numpy as np

# Створення одновимірного масиву з 10 випадкових чисел від 0 до 100
arr = np.random.randint(0, 101, size=10)
print("Вихідний масив:")
print(arr)

# Обчислення мінімального, максимального та середнього значень
min_val = np.min(arr)
max_val = np.max(arr)
mean_val = np.mean(arr)

print(f"\nМінімальне значення: {min_val}")
print(f"Максимальне значення: {max_val}")
print(f"Середнє значення: {mean_val:.2f}")

# Сортування масиву за зростанням
sorted_arr = np.sort(arr)

print("\nМасив до сортування:")
print(arr)
print("Масив після сортування:")
print(sorted_arr)
```

```python
# Створення двовимірного масиву 4x4 з випадковими цілими числами від 1 до 50
matrix = np.random.randint(1, 51, size=(4, 4))

print("Матриця 4x4:")
print(matrix)

# Сума за рядками (axis=1) та за стовпцями (axis=0)
row_sums = np.sum(matrix, axis=1)
col_sums = np.sum(matrix, axis=0)

# Загальне середнє значення
total_mean = np.mean(matrix)

print(f"\nСуми за рядками: {row_sums}")
print(f"Суми за стовпцями: {col_sums}")
print(f"Загальне середнє значення: {total_mean:.2f}")
```
